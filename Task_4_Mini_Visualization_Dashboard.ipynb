```python
#Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
#Load Dataset

df=pd.read_csv("train.csv")
df.head()
```

```python
#Explore Dataset

df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
#Data Cleaning

df["Age"]=df["Age"].fillna(df["Age"].mean())
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].
mode()[0])

df.drop("Cabin",axis=1,inplace=True)

df.isnull().sum()
```

```python
#Fanily Size

df["Family_Size"]=df["SibSp"]+df["Parch"]+1
```

```python
#Age Group

bins=[0,12,18,35,60,100]

labels=["Child","Teen","Adult","Middle Age","Senior"]

df["Age_Group"]=pd.cut(df["Age"],bins=bins,labels=labels)
```

```python
#Dashboard

#Chart 1:Survival Count

plt.figure(figsize=(6,4))
sns.countplot(x="Survived",data=df)
plt.title("Survival Count")
plt.show()
```

```python
#Chart 2:Survival By Gender

plt.figure(figsize=(6,4))
sns.countplot(x="Sex",hue="Survived",data=df)
plt.title("Survival By Gender")
plt.show()
```

```python
#Chart 3:Survival by passenger class

plt.figure(figsize=(6,4))
sns.countplot(x="Pclass",hue="Survived",data=df)
plt.title("Survival by passenger class")
plt.show()
```

```python
#Chart 4:Age Distribution

plt.figure(figsize=(7,4))
sns.histplot(x="Age",data=df,kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
#Chart 5:Survival by age group

plt.figure(figsize=(8,4))
sns.barplot(x="Age_Group",hue="Survived",data=df)
plt.title("Survival by age group")
plt.show()
```

```python
#Chart 6:Family size distribution

plt.figure(figsize=(8,4))
sns.countplot(x="Family_Size",data=df)
plt.title("Family size distribution")
plt.show()
```

```python
#Chart 7:Correlation Heatmap

plt.figure(figsize=(10,7))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
